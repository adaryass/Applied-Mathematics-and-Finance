```python
import numpy as np
from numpy.random import normal 
import matplotlib.pyplot as plt
```

```python
def generate_process(dt = 0.1, theta = 1.2, mu = 0.5, sigma = 0.3, n = 10000):
    # r (t =0) = 0 et l'initialization de r(t) avec 0
    r = np.zeros(n)

    for t in range(1, n):
        r[t] = r[t-1] + theta * (mu - r[t-1]) * dt + sigma * normal(0, np.sqrt(dt))
    return r

def plot_process(r):
    plt.plot(r)
    plt.xlabel('t')
    plt.ylabel('r(t)')
    plt.title('Ornstein-Uhlenbeck Process')
    plt.show()

if __name__ == "__main__":
    data = generate_process()
    plot_process(data)
```
